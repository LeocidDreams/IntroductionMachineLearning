```python
#dropp student_id and course columns
#sort current dataframe by age
#check NAN values
def dataCleaning(df:pd.DataFrame):
    df=df.T.drop(['student_id','course'],axis=0)
    df=df.T.sort_values(by='age',ascending=True,inplace=False)
    df=df[(df.T!='NAND').any()] #instead of .isna()
    print(df)
    print(df.columns)
    print(df.index)
    return df

# print(df.index)
# print(df.columns)
# print(df.columns)
# print(dataCleaning(df))
# dataCleaning(df)
# print(df)

# #dropping student_id and course columns
# print(df.columns)
# df=df.T.drop(['student_id','course'],axis=0)
# print(df.index)
# #sorting columns by age
# df=df.T.sort_values(by='age',ascending=False,inplace=False)
# print(df)


```
