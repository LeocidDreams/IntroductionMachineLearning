```python
##Predicting car price
year_price_dataFrame=pd.DataFrame(
    {'Year':new_dataFrame['Year'],'Price':new_dataFrame['Price']},
)

#creating  plot
# plt.plot(new_dataFrame['Year'].mean(),new_dataFrame['Price'])

#cleaning missing values
year_price_dataFrame.dropna(inplace=True)

group=year_price_dataFrame.groupby('Year')['Price'].mean().plot(kind='line')
plt.show()
```

```python

##splitting test data and train data
y=year_price_dataFrame['Price'].to_numpy().astype(np.int32).reshape(-1,1)
X=year_price_dataFrame['Year'].to_numpy().reshape(-1,1)
# print(X)
# print(y)


X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)

```

```python
#assigning choosen model
lin_reg=Ridge()
lin_reg.fit(X_train,y_train)
```

```python
#predicting model
pred=lin_reg.predict(X_test)
print(pred)
#choosen validation method
##mean_absolute error
mae=mean_absolute_error(y_test,pred)
print(f'Mean absolute error: {mae:3.3} ({mae/np.mean(pred)*100:3.3}%)')
#mean square error
# mse=mean_squared_error(y_test,pred)
# print(f'Mean square error:{mse} ({mse/np.mean(pred)})')
#rmsd
rmsd=root_mean_squared_error(y_test,pred)
print(f'root mean squared error: {rmsd} ({rmsd/np.mean(pred)*100:3.3}%)')
```

```python
##coefficient determination(r^2)

score=lin_reg.score(X_train,y_train)
print(f'Linear regression score: {score*100:3.3}%')
```
