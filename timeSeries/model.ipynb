```python
#importing required libraries
import pandas as pd
import numpy as np
import os
from collections import UserDict

#importing required dataset
energy=pd.read_csv('energy.csv',parse_dates=['timestamp'],index_col=['timestamp'])
# print(energy.columns)
# print(energy.index)

#Reindex the dataframe between the minimum and maximum timestamp in the time series.

# energy.index=energy['timestamp']
# energy=energy.reindex(
#     pd.date_range(
#         start=min(energy['timestamp']),
#         end=max(energy['timestamp']),
#         freq='H'
#     )
# )
energy=energy.reindex(
    pd.date_range(
        start=energy.index.min(),
        end=energy.index.max(),
        freq='h'
    ),
   
)
# print(energy)
# energy=energy.drop('timestamp',axis=1)
energy=energy['load']
energy.head()

```

```python
%matplotlib inline
import matplotlib.pyplot as plt
energy.plot(
    y='load',
    subplots=True,
    figsize=(15,6),
    fontsize=12
)
plt.xlabel('timestamp',fontsize=12)
plt.ylabel('load',fontsize=12)
plt.show()
```
